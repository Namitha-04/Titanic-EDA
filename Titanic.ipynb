```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

data.groupby('Pclass')['Age'].mean().plot(ax=axes[0,0], kind='line', marker='o', color='steelblue')
axes[0,0].set_title('Average Age by Class')
axes[0,0].set_xlabel('Passenger Class')
axes[0,0].set_ylabel('Age')

data['Survived'].value_counts().plot(ax=axes[0,1], kind='bar', color=['salmon', 'steelblue'])
axes[0,1].set_title('Survival Count')
axes[0,1].set_xticklabels(['Died', 'Survived'], rotation=0)

axes[1,0].scatter(data['Age'], data['Fare'], alpha=0.4, color='purple')
axes[1,0].set_title('Age vs Fare')
axes[1,0].set_xlabel('Age')
axes[1,0].set_ylabel('Fare')

axes[1,1].hist(data['Age'].dropna(), bins=20, color='teal', edgecolor='white')
axes[1,1].set_title('Age Distribution')
axes[1,1].set_xlabel('Age')
axes[1,1].set_ylabel('Count')
```

```python
plt.scatter(data['Pclass'], data['Age'])
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.title('Scatter Plot of Passenger Class vs Age')
```

```python
plt.figure(figsize=(8,5))
sns.violinplot(data, x='Pclass', y='Fare', hue='Survived', split=True, palette='muted')
plt.title('Fare Distribution by Class and Survival')
plt.savefig('plot4.png'); plt.show()
```
